### 平均

```r
# データ x = (9, 5, 10, 11, 1) に対して，平均を計算してみよう．

# データ
x = c(9, 5, 10, 11, 1)

# Q.1 平均を mean 関数を用いて計算しよう．
print(c('Q.1', mean(x)))

# Q.2 平均を手で計算しよう．
# (9 + 5 + 10 + 11 + 1) / 5 = 36 / 5 = 7.2
print(c('Q.2', 7.2))

# Q.3 R で平均を四則演算子のみを用いて計算しよう（for を使いません）．
m = (9 + 5 + 10 + 11 + 1) / 5
print(c('Q.3', m))

# Q.4 R で平均を sum 関数と四則演算子を用いて計算しよう．
m = sum(x) / 5
print(c('Q.4', m))

# Q.5 R で平均を四則演算子と for ループを用いて計算しよう．
s = 0
for (i in 1:5) {
  s = s + x[i]
}
m = s / 5
print(c('Q.5', m))
```

### 不偏分散

```r
# データ x = (9, 5, 10, 11, 1) に対して，不偏分散を計算してみよう．

# データ
x = c(9, 5, 10, 11, 1)

# Q.1 不偏分散を var 関数を用いて計算しよう．
print(c('Q.1', var(x)))

# Q.2 不偏分散を手で計算しよう．
# 平均 m = 7.2
# ((9-7.2)^2 + (5-7.2)^2 + (10-7.2)^2 + (11-7.2)^2 + (1-7.2)^2) / (5-1)
# = (3.24 + 4.84 + 7.84 + 14.44 + 38.44) / 4 = 68.8 / 4 = 17.2
print(c('Q.2', 17.2))

# Q.3 R で不偏分散を四則演算子のみを用いて計算しよう（for を使いません）．
m = (9 + 5 + 10 + 11 + 1) / 5
v = ((9 - m)^2 + (5 - m)^2 + (10 - m)^2 + (11 - m)^2 + (1 - m)^2) / (5 - 1)
print(c('Q.3', v))

# Q.4 R で不偏分散を sum 関数と mean 関数と四則演算子を用いて計算しよう．
m = mean(x)
v = sum((x - m)^2) / (5 - 1)
print(c('Q.4', v))

# Q.5 R で不偏分散を四則演算子と for ループを用いて計算しよう．
s = 0
for (i in 1:5) {
  s = s + x[i]
}
m = s / 5
ss = 0
for (i in 1:5) {
  ss = ss + (x[i] - m)^2
}
v = ss / (5 - 1)
print(c('Q.5', v))
```

### 偏りのある分散

```r
# データ x = (9, 5, 10, 11, 1) に対して，偏りのある分散を計算してみよう．

# データ
x = c(9, 5, 10, 11, 1)

# Q.1 偏りのある分散を var 関数を用いて計算しよう．
print(c('Q.1', var(x) * 4 / 5))

# Q.2 偏りのある分散を手で計算しよう．
# 平均 m = 7.2
# ((9-7.2)^2 + (5-7.2)^2 + (10-7.2)^2 + (11-7.2)^2 + (1-7.2)^2) / 5
# = 68.8 / 5 = 13.76
print(c('Q.2', 13.76))

# Q.3 R で偏りのある分散を四則演算子のみを用いて計算しよう（for を使いません）．
m = (9 + 5 + 10 + 11 + 1) / 5
v = ((9 - m)^2 + (5 - m)^2 + (10 - m)^2 + (11 - m)^2 + (1 - m)^2) / 5
print(c('Q.3', v))

# Q.4 R で偏りのある分散を sum 関数と mean 関数と四則演算子を用いて計算しよう．
m = mean(x)
v = sum((x - m)^2) / 5
print(c('Q.4', v))

# Q.5 R で偏りのある分散を四則演算子と for ループを用いて計算しよう．
s = 0
for (i in 1:5) {
  s = s + x[i]
}
m = s / 5
ss = 0
for (i in 1:5) {
  ss = ss + (x[i] - m)^2
}
v = ss / 5
print(c('Q.5', v))
```

### 不偏共分散

```r
# データ x = (9, 5, 10, 11, 1)
#      y = (5, 20, 1, 12, 4) に対して，不偏共分散を計算してみよう．

# データ
x = c(9, 5, 10, 11, 1)
y = c(5, 20, 1, 12, 4)

# Q.1 不偏共分散を cov 関数を用いて計算しよう．
print(c('Q.1', cov(x, y), cov(x, y) * 4 / 5))

# Q.2 不偏共分散を手で計算しよう．
# mx = 7.2, my = 42/5 = 8.4
# (1.8*(-3.4) + (-2.2)*11.6 + 2.8*(-7.4) + 3.8*3.6 + (-6.2)*(-4.4)) / 4
# = (-6.12 - 25.52 - 20.72 + 13.68 + 27.28) / 4 = -11.4 / 4 = -2.85
print(c('Q.2', -2.85))

# Q.3 R で不偏共分散を四則演算子のみを用いて計算しよう（for を使いません）．
mx = (9 + 5 + 10 + 11 + 1) / 5
my = (5 + 20 + 1 + 12 + 4) / 5
c = ((9 - mx) * (5 - my) + (5 - mx) * (20 - my) + (10 - mx) * (1 - my) + (11 - mx) * (12 - my) + (1 - mx) * (4 - my)) / (5 - 1)
print(c('Q.3', c))

# Q.4 R で不偏共分散を sum 関数と mean 関数と四則演算子を用いて計算しよう．
c = sum((x - mean(x)) * (y - mean(y))) / (5 - 1)
print(c('Q.4', c))

# Q.5 R で不偏共分散を四則演算子と for ループを用いて計算しよう．
mx = mean(x)
my = mean(y)
ss = 0
for (i in 1:5) {
  ss = ss + (x[i] - mx) * (y[i] - my)
}
c = ss / (5 - 1)
print(c('Q.5', c))
```

### 相関係数

```r
# データ x = (9, 5, 10, 11, 1)
#      y = (5, 20, 1, 12, 4) に対して，相関係数を計算してみよう．

# データ
x = c(9, 5, 10, 11, 1)
y = c(5, 20, 1, 12, 4)

# Q.1 相関係数を cor 関数を用いて計算しよう．
print(c('Q.1', cor(x, y)))

# Q.2 相関係数を手で計算しよう．
# cov(x,y) = -2.85, var(x) = 17.2, var(y) = 58.3
# r = -2.85 / (sqrt(17.2) * sqrt(58.3)) = -2.85 / 31.66 ≈ -0.09
print(c('Q.2', -2.85 / (sqrt(17.2) * sqrt(58.3))))

# Q.3 R で相関係数を四則演算子のみを用いて計算しよう（for を使いません）．
mx = (9 + 5 + 10 + 11 + 1) / 5
my = (5 + 20 + 1 + 12 + 4) / 5
c = ((9 - mx) * (5 - my) + (5 - mx) * (20 - my) + (10 - mx) * (1 - my) + (11 - mx) * (12 - my) + (1 - mx) * (4 - my)) / (5 - 1)
vx = ((9 - mx)^2 + (5 - mx)^2 + (10 - mx)^2 + (11 - mx)^2 + (1 - mx)^2) / (5 - 1)
vy = ((5 - my)^2 + (20 - my)^2 + (1 - my)^2 + (12 - my)^2 + (4 - my)^2) / (5 - 1)
r = c / sqrt(vx) / sqrt(vy)
print(c('Q.3', r))

# Q.4 R で相関係数を sum 関数と mean 関数と四則演算子を用いて計算しよう．
c  = sum((x - mean(x)) * (y - mean(y))) / (5 - 1)
vx = sum((x - mean(x)) * (x - mean(x))) / (5 - 1)
vy = sum((y - mean(y)) * (y - mean(y))) / (5 - 1)
r = c / sqrt(vx) / sqrt(vy)
print(c('Q.4', r))

# Q.5 R で相関係数を cov 関数と sd 関数を用いて計算しよう．
r = cov(x, y) / sd(x) / sd(y)
print(c('Q.5', r))
```

### 回帰係数

```r
# データ x = (9, 5, 10, 11, 1)
#      y = (5, 20, 1, 12, 4) に対して，回帰係数を計算してみよう．

# データ
x = c(9, 5, 10, 11, 1)
y = c(5, 20, 1, 12, 4)

# Q.1 R の lm モデル式を用いて計算しよう．
model = lm(y ~ x)
print(c('Q.1', model$coefficients))

# Q.2 回帰係数を手で計算しよう．
# b = cov(x,y) / var(x) = -2.85 / 17.2 ≈ -0.1657
# a = my - b * mx = 8.4 - (-0.1657) * 7.2 ≈ 9.593
print(c('Q.2 a', 8.4 - (-2.85 / 17.2) * 7.2))
print(c('Q.2 b', -2.85 / 17.2))

# Q.3 R で回帰係数を四則演算子のみを用いて計算しよう（for を使いません）．
mx = (9 + 5 + 10 + 11 + 1) / 5
my = (5 + 20 + 1 + 12 + 4) / 5
cov_xy = ((9 - mx) * (5 - my) + (5 - mx) * (20 - my) + (10 - mx) * (1 - my) + (11 - mx) * (12 - my) + (1 - mx) * (4 - my)) / (5 - 1)
var_x  = ((9 - mx)^2 + (5 - mx)^2 + (10 - mx)^2 + (11 - mx)^2 + (1 - mx)^2) / (5 - 1)
b = cov_xy / var_x
a = my - b * mx
print(c('Q.3', a, b))

# Q.4 R で回帰係数を sum 関数と mean 関数と四則演算子を用いて計算しよう．
mx = mean(x)
my = mean(y)
b = sum((x - mx) * (y - my)) / sum((x - mx)^2)
a = my - b * mx
print(c('Q.4', a, b))

# Q.5 R で回帰係数を cor 関数と sd 関数と mean 関数を用いて計算しよう．
b = cor(x, y) * sd(y) / sd(x)
a = mean(y) - b * mean(x)
print(c('Q.5', a, b))
```
